## Phase 6: Check your progress

Run the cell below. It asks Git for your commit history and tells you which phases are visible in it. It only reads; it does not change anything.

````python
import subprocess                                                    # lets Python run a program on the computer

result = subprocess.run(['git', 'log', '--format=%s'], capture_output=True, text=True)   # ask Git for the commit messages
messages = result.stdout.splitlines()                                # one message per line

print('Commits in this repository:', len(messages))                 # how many commits so far
for message in messages:                                             # show them, newest first
    print('  -', message)

expected = ['Add practice notebook', 'Fix Tuesday count', 'Add scrap rate', 'Add teammate name']   # one per phase
print()
for number, message in enumerate(expected, start=1):                 # check each expected message
    if message in messages:                                          # is it in the history?
        print(f'Phase {number}: done   ({message})')
    else:
        print(f'Phase {number}: not yet ({message})')
````
